```python
from ai import MAB_Agent, Epsilon_Greedy_MAB_Agent

egreedy_agents = []
for i in range(5):
    epsilon = i / 4
    egreedy_agent = MAB_Agent.load(f"../data/egreedy_agent(e={epsilon})")
    egreedy_agents.append(egreedy_agent)
    
for step in [250, 500, 750, 1000]:
    print(f"step {step:4}")
    for egreedy_agent in egreedy_agents:
        epsilon = egreedy_agent.epsilon
        print(f"epsilon = {epsilon:.2f}:",
              f"1 ステップあたりに稼ぐコインの平均枚数 {egreedy_agent.mean_mean_coins_by_step[step]:.3f}",
              f"理論値 {epsilon * 0.5 + (1 - epsilon) * 0.7:.3f}")
```
