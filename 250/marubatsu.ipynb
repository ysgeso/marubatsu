```python
from ai import MAB_Agent, Random_MAB_Agent
import numpy as np

class Greedy_MAB_Agent(MAB_Agent):
    NAME = "貪欲法"
    
    def select_arm(self):
        maxcoin = max(self.slot["mean_coin"])
        bestarms = []
        for arm in range(self.mab.num):
            if maxcoin == self.slot["mean_coin"][arm]:
                bestarms.append(arm)           
        return np.random.choice(bestarms)
```

```python
p = [0.3, 0.5, 0.7]
random_agent = Random_MAB_Agent(p=p)
random_agent.play_episodes(stepnum=10000, episodenum=1000)
random_agent.save("../data/random_agent")
greedy_agent = Greedy_MAB_Agent(p=p)
greedy_agent.play_episodes(stepnum=10000, episodenum=1000)
greedy_agent.save("../data/greedy_agent")
```
