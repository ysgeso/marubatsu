```python
import matplotlib.pyplot as plt
import numpy as np
import japanize_matplotlib

S = np.linspace(0, 1000, 1000)
N = 1000 - S
plt.xlabel("販売価格")
plt.ylabel("販売個数")
plt.plot(S, N)
```

```python
P = -S ** 2 + 1100 * S - 100000
plt.xlabel("販売価格")
plt.ylabel("利益")
plt.plot(S, P)
```

```python
N = (1000 - S) ** 2 / 1000
plt.xlabel("販売価格")
plt.ylabel("販売個数")
plt.plot(S, N)
```

```python
for i in range(1, 6):
    N = (1000 - S) ** i / (1000 ** (i - 1))
    plt.xlabel("販売価格")
    plt.ylabel("販売個数")
    plt.plot(S, N, label=f"i = {i}")
plt.legend()
```

```python
fig, axes = plt.subplots(2, 2)

for n in range(1, 5):
    ax = axes[(n - 1) // 2][(n - 1) % 2]
    X = np.linspace(-0.1, n - 0.9, 100)
    Y = X
    for i in range (1, n):
        Y = Y * (X - i)
    ax.set_title(f"{n} 次関数")
    ax.plot(X, Y)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 3))

ax = axes[0]
X = np.linspace(-10, 10, 100)
Y = 1 / (1 + X ** 2)
ax.set_title("y = 1 / (1 + x * x)")
ax.plot(X, Y)

ax = axes[1]
X = np.linspace(-8, 8, 100)
Y = np.sin(X)
ax.set_title("y = sin(x)")
ax.plot(X, Y)

ax = axes[2]
X = np.linspace(-1, 1, 100)
Y = np.sqrt(1 - X ** 2)
ax.set_title("y = √(1 - x * x)")
ax.plot(X, Y)

plt.tight_layout()
plt.show()
```
