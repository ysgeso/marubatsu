```python
from ai import MAB, MAB_Agent
import numpy as np

class Epsilon_Greedy_MAB_Agent(MAB_Agent):
    NAME = "ε-greedy 法"

    def __init__(self, p, epsilon):
        self.mab = MAB(p=p)
        self.epsilon = epsilon
        self.title = f"{self.NAME}(ε={epsilon})"
        
    def select_arm(self):
        if np.random.uniform() < self.epsilon:
            return np.random.choice(self.mab.num)
        else:
            maxcoin = max(self.slot["mean_coin"])
            bestarms = []
            for arm in range(self.mab.num):
                if maxcoin == self.slot["mean_coin"][arm]:
                    bestarms.append(arm)           
            return np.random.choice(bestarms)
```

```python
egreedy_agents = []

p = [ 0.3, 0.5, 0.7 ]
for i in range(5):
    epsilon = i / 4
    egreedy_agent = Epsilon_Greedy_MAB_Agent(p=p, epsilon=epsilon)
    egreedy_agent.play_episodes(stepnum=10000, episodenum=1000)
    egreedy_agent.save(f"../data/egreedy_agent(e={epsilon})")
    egreedy_agents.append(egreedy_agent)
```
